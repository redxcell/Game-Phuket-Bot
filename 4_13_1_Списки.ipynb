### **Списки**

Вот мы и добрались до типов языка python, которые представляют собой *контейнеры*, то есть, это типы данных, которые являются хранилищами для других типов данных.

Один из таких типов -- `list` (массив, список, лист). Это контейнер, куда можно положить сколько угодно других переменных, значений, и эти переменные даже могут быть разных типов и даже сами могут быть контейнерами! То есть, в `list` можно положить `list`.

Чтобы задать `список`, надо в квадратные скобки `[]` записать нужные элементы. Пустые скобки задают пустой `list`. Пустой лист также можно задать, написав `list()`:

```python
list1 = list()
list2 = []
```

Что же сейчас там внутри этих списков? Выведем их в консоль.

```python
print(list1)
print(list2)
```

Пока пусто. Как же занести туда данные? Значения списка можно задать перечислением его элементов. Отдельные элементы списка разделяются запятой, а весь список заключается в квадратные скобки:


```python
list1 = [1, 2, 3, 4, 5]
list2 = ["Пушкин","Толстой","Маяковский","Есенин","Горький", "Гайдар"]
```

Посмотрим, как выглядят наши списки теперь:

```python
print(list1)
print(list2)
print(list1, list2)
```

Мы увидели списки двух типов: чисел и текста.
Но вообще элементы одного списка могут быть разного типа:

```python
listPeople = ["Билл Гейтс", "William Henry Gates III", 1955, "США", "Сиэтл", "microsoft.com"]
print(listPeople)
```

## Операции со списками


### Индексация

```python
list2 = ["Пушкин","Толстой","Маяковский","Есенин","Горький"]
```

```python
# Первый элемент имеет нулевой индекс
print(list2[0])
```

```python
# Второй элемент - индекс 1
print(list2[1])
```

```python
# Последний элемент, индекс -1
print(list2[-1])
```

```python
# ПРЕДпоследний элемент, индекс -2
print(list2[-2])
```

```python
# Ошибка т.к. вышли за пределы списка - list index out of range
print(list2[5])
```

Выведем все элементы списка по очереди:

```python
for i in list2:
  print(i)
```

### Изменение элементов списка

```python
# Назначение элементу другого значения:
list1 = [1, 2, 3, 4, 5]
list1[0] = 100
print(list1)
```

```python
# Увеличение значения элемента:
list1[0] = list1[0] + 1
print(list1)
```

### Сложение списков (конкатенация)

Списки можно считать обобщением строк в том смысле, что списки - это тоже упорядоченные последовательности, но значения списка могут быть произвольного типа. Большинство операций для строк работают и для списков:

### Разделение на части

При создании списков, содержащих строковые величины, можно использовать метод split (разделить на части).

Этот метод применяется к строковым объектам и позволяет разбивать их на элементы, разделенные заданным символом.

Каждый элемент после применения метода split становится отдельным элементом нового списка.

```python
animals = 'кошка,собака,хомяк,морская свинка,попугай,лошадь'
```

```python
list_animals = animals.split(",")
```

```python
print(list_animals)
```

```python
print(list_animals[0])
```

```python
#Cто 45 пять 6733 42 восемь
```

```python
numbers = input("Введи свои любимые числа через пробел и нажми Enter: ")
numbers = numbers.split(" ")

print(numbers)
```

Упростим такую структуру, применив .split сразу при вводе:

```python
numbers = input("Введи свои любимые числа через пробел и нажми Enter: ").split(" ")
print(numbers)
```

### Сортировка

1. Метод **list.sort()** — самый простой и часто используемый способ .сортировки списков в Python. Его ключевая особенность в том, что он модифицирует исходный список напрямую, не создавая копий. Это делает его более эффективным по памяти, но требует осторожности, если вам нужно сохранить оригинальный порядок элементов.

Базовый синтаксис метода прост:

```python
myList = [3, 1, 4, 1, 5, 9, 2, 6]
```

```python
print(myList)
```

```python
# Отсортируем список по возрастанию
myList2 = myList + []
myList2.sort()
```

```python
print(myList)
print(myList2)
```

Для сортировки в обратном порядке используется **параметр reverse=True**:

```python
myList.sort(reverse=True)
```

```python
print(myList)
```

2. Ещё раз подчеркну, что метод sort() сортирует именно исходный список. Результат сортировки никуда нельзя присвоить - список уже отсортирован.

Если нужен новый остортированный список на базе исходного, используется другой метод - sorted(). Он как раз возвращает новый отсортированный список. Это делает его более универсальным и безопасным для использования в ситуациях, когда важно сохранить исходный порядок элементов.

```python
myList = [3, 1, 4, 1, 5, 9, 1, 3, 6, 5, 0, 1, 13]
```

```python
print(myList)
```

```python
# Сортируем и записываем результат в другой список
sortedList = sorted(myList)
```

```python
print(myList)
print(sortedList)
```

3. Функция sorted() тоже принимает параметр reverse для сортировки в обратном порядке:

```python
sortedList = sorted(myList, reverse=True)
```

```python
print(myList)
print(sortedList)
```

4. Сортировка строк без учета регистра. Это еще один параметр:

```python
fruits = ["Вишня", "ананас", "ЯБЛОКО", "груша"]
```

```python
sortedFruits = sorted(fruits, key=str.lower)
```

```python
print(sortedFruits)
```

На самом деле, ключей для настройки сортировки гораздо больше, их без труда можно найти в интернетах, а мы не будем на них останавливаться подробнее.

### Поиск в списках

1. Метод index() — возвращает индекс первого вхождения элемента

Базовый синтаксис метода index():

list.index(element[, start[, end]])

```python
myList = [3, 1, 4, 1, 5, 9, 1, 3, 6, 5, 0, 1, 13]
```

```python
print(myList)
```

```python
print(myList.index(1))
```

```python
# Поищем 1, начиная со 2 элемента:
print(myList.index(1, 2))
```

При больших списках index() может быть неэффективным и тяжелым, так как перебирает в лоб весь список, пока не наткнется на нужный элемент.

2. Оператор in. Проверяет наличие элемента в списке.

Возвращает False, если такого не найдено или True, если таковой есть.

```python
myList = [3, 1, 4, 1, 5, 9, 1, 3, 6, 5, 0, 1, 13]
```

```python
print(1 in myList)
```

3. Метод count() — это простой, но мощный инструмент для подсчета количества вхождений определенного элемента в список.

```python
print(myList.count(1))
```

```python
print(myList.count(3))
```

```python
print(myList.count(100))
```

Метод count() особенно полезен в следующих сценариях:

*   Анализ частоты элементов в наборе данных
*   Проверка уникальности элементов
*   Фильтрация данных на основе их распространенности
*   Построение простых гистограмм и визуализаций
*   Валидация данных и контроль качества информации

При работе с большими списками следует помнить, что метод count() имеет линейную сложность O(n), т.е. время выполнения пропорционально размеру списка.
